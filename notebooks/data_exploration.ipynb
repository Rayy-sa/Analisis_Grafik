```python
import os
import json
import matplotlib.pyplot as plt
from PIL import Image
from glob import glob

# Tentukan path utama
raw_data_dir = '../data/raw/'
natural_dir = os.path.join(raw_data_dir, 'natural_data')
synthetic_dir = os.path.join(raw_data_dir, 'synthetic_data')

# ==========================================
# 1. EKSPLORASI NATURAL DATA
# ==========================================
print("=== EKSPLORASI NATURAL DATA ===")
nat_img_dir = os.path.join(natural_dir, 'images')
nat_ann_dir = os.path.join(natural_dir, 'annotations')
nat_jsn_dir = os.path.join(natural_dir,'index.json')

# Ambil semua file gambar
nat_images = glob(os.path.join(nat_img_dir, '*.*'))
print(f"Total gambar natural: {len(nat_images)}")

if len(nat_images) > 0:
    # Ambil sampel pertama
    sample_nat = nat_images[0]
    base_name = os.path.splitext(os.path.basename(sample_nat))[0]
    
    # Tampilkan Gambar
    img = Image.open(sample_nat)
    plt.figure(figsize=(8, 5))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Natural Data: {os.path.basename(sample_nat)}")
    plt.show()
    
print("\n" + "="*50 + "\n")

# ==========================================
# 2. EKSPLORASI SYNTHETIC DATA
# ==========================================
print("=== EKSPLORASI SYNTHETIC DATA ===")
syn_chart_dir = os.path.join(synthetic_dir, 'charts')
syn_md_dir = os.path.join(synthetic_dir, 'markdown')

# Ambil semua file gambar
syn_images = glob(os.path.join(syn_chart_dir, '*.*'))
print(f"Total gambar synthetic: {len(syn_images)}")

if len(syn_images) > 0:
    # Ambil sampel pertama
    sample_syn = syn_images[0]
    base_name = os.path.splitext(os.path.basename(sample_syn))[0]
    
    # Tampilkan Gambar
    img = Image.open(sample_syn)
    plt.figure(figsize=(8, 5))
    plt.imshow(img)
    plt.axis('off')
    plt.title(f"Synthetic Data: {os.path.basename(sample_syn)}")
    plt.show()
    
    # Cari anotasi Markdown yang pasangannya
    md_path = os.path.join(syn_md_dir, f"{base_name}.md")
    if os.path.exists(md_path):
        with open(md_path, 'r', encoding='utf-8') as f:
            md_content = f.read()
            print("Ground Truth (Markdown):")
            # Cetak 500 karakter pertama
            print(md_content[:500] + "\n... [dipotong]\n")
    else:
        print("File Markdown pasangan tidak ditemukan.")
```
